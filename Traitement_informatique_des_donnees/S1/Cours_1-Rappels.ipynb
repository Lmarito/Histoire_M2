## Deuxième partie : Faisons du Python

### Des éléments en mémoire
Python, pour fonctionner, va aller chercher des éléments stockés en mémoire de votre ordinateur. Il peut s'agir de :
* **bibliothèques (_libraries_)** :
    * Ensemble de code qui va permettre aux utilisateurs d'aller "plus vite" dans les démarches. Par exemple, si une bibliothèque permettait d'ouvrir une porte dans la vraie vie, on aurait quelque chose comme :    

    ```python
    door.open()
    ```     

    plutôt que :    

    ```python 
    poignee = door.trouver_la_poignée()    
    poignee_saisie = poignee.saisir()    
    poignee_clenchee = poignee_saisie.appuyer()    
    door.pousser()    
    ```    

* **Variables** :
    * Ensemble d'éléments sur lesquels l'utilisateur va pouvoir agir.une variable est _affectée_, par le biais de l'opérateur "=". Il peut s'agir du résultat d'un calcul sur x autres variables, ou d'une affectation directe :
    ```python
    variable = 1 # Affectation directe
    variable_2 = 2
    variable_3 = variable_2 - variable # Affectation par calcul
    ```
    * **Une variable ne peut pas contenir d'espace, et elle s'écrit avec des lettres, des chiffres et des underscore (8) `_`**. Par convention, on nommme une variable sans majuscules.


### L'algorithmique : une recette de cuisine

Un algorithme, c'est comme une recette : une suite d'instructions à suivre dans l'ordre.

**Exemple : calculer une moyenne de notes**

````python
# 1. Trouver les notes 
note_1 = 14
note_2 = 12
note_3 = 19
note_4 = 6
note_5 = 17
note_6 = 3

# 2. Les additionner
total = note_1 + note_2 + note_3 + note_4 + note_5 + note_6
print("Le total est de", total)

# 3. Repérer le nombre de notes
# On sait qu'il y a 6 notes ici. On va donc créer une variable "nb" qui contient cette information
nb = 6

# 4. Diviser la somme par le nombre de notes
moyenne = total / nb
print("La moyenne est de", moyenne)
````

**Concept clé : les variables**

Une **variable** est comme une boîte avec une étiquette. On y met une valeur qu'on peut réutiliser.
- `note_xx` est une série de variables qui contiennent des **nombres**
- `total` est une variable qui contient la somme
- On utilise `=` pour **affecter** une **valeur** à une **variable**

### 1. Les nombres

**`float` (décimaux) :**

````python
taille = 1.75
temperature = 20.5
taux = 0.15
type(taille)  # Renvoie : float
````

### 2. Les textes

**`str` (chaînes de caractères) :**

````python
prenom = "Marie"
ville = 'Paris'
phrase = "Bonjour à tous"
type(prenom)  # Renvoie : str
````

Attention : `"42"` (avec guillemets) est du texte, pas un nombre !
**Les double guillemets `"` (3) et les guillemets simples `'` (4) permettent de déclarer une chaîne de caractères.**    
Il faut cependant **absolument** que le guillemet ouvrant soit le même que le guillemet fermant.     
Pour déclarer des guillemets au sein de la chaîne de caractères, on les **échappe** :     
On met alors un antislash `\` avant le caractère en question : 

````python
# Sans l'échappement → erreur :

test_echec = "ceci est un guillemet : ""
print(test_echec)
````

````python
# Avec l'échappement → correct :
test = "ceci est un guillemet : \""
print(test)
````

**L'échappement des caractères est un concept assez fondamental en Python, on le verra probablement plus tard !**
